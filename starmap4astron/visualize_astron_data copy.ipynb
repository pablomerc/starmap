````python
import numpy as np
import matplotlib.pyplot as plt

# Path to your .npz file
file_path = '/Users/pablom.perez/Desktop/MIT-PhD-macbook/starmap/starmap4astron/synthetic_data_output/dataset2/synthetic_starry_data_part_0.npz'

# Load the data
data = np.load(file_path, allow_pickle=True)
print("Available keys:", list(data.keys()))

# Extract arrays
time = data['time']         # shape: (n_examples, N_PTS)
flux = data['flux']         # shape: (n_examples, N_PTS)
y_coeffs = data['y_coeffs'] # shape: (n_examples, coeff_count)
params = data['parameters'] # shape: (n_examples,)

# Plot the first example's light curve
plt.figure()
plt.plot(time[0], flux[0])
plt.xlabel('Time')
plt.ylabel('Flux')
plt.title('Synthetic Light Curve - Example 0')
plt.show()

# Visualize the corresponding image (if available)
if 'image' in data:
    img = data['image'][0]
    plt.figure()
    plt.imshow(img, origin='lower')
    plt.title('Rendered Map - Example 0')
    plt.axis('off')
    plt.show()
else:
    print('No image data found.')
```
````
