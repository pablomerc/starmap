```python
#!/usr/bin/env python3
import os
import glob
from PIL import Image
import matplotlib.pyplot as plt

# Base directory containing all your lr* folders
BASE_DIR = "/Users/pablom.perez/Desktop/MIT-PhD-macbook/starmap/starmap4astron/hpam_tuning_output"

# Glob for all folders starting with "lr"
pattern = os.path.join(BASE_DIR, "lr*")



for folder in sorted(glob.glob(pattern)):
    # print('hello')
    in_path   = os.path.join(folder, "example", "sample_input.png")
    recon_path = os.path.join(folder, "example", "sample_recon.png")

    # Skip if either image is missing
    if not (os.path.isfile(in_path) and os.path.isfile(recon_path)):
        print(f"Skipping {folder} - missing images")
        continue

    # Load images
    img_in  = Image.open(in_path)
    img_re  = Image.open(recon_path)

    # Plot side-by-side
    fig, axes = plt.subplots(1, 2, figsize=(8, 4))
    fig.suptitle(os.path.basename(folder), fontsize=14)

    axes[0].imshow(img_in,cmap="plasma")
    axes[0].set_title("Input")
    axes[0].axis("off")

    axes[1].imshow(img_re,cmap="plasma")
    axes[1].set_title("Reconstruction")
    axes[1].axis("off")

    plt.tight_layout()
    plt.show()
```
