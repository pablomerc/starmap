```python
import numpy as np
import matplotlib.pyplot as plt

# Path to your .npz file
file_path = '/Users/pablom.perez/Desktop/MIT-PhD-macbook/starmap/starmap4astron/synthetic_data_output/dataset2/synthetic_starry_data_part_0.npz'

# Load the data
data = np.load(file_path, allow_pickle=True)
print("Available keys:", list(data.keys()))

# Extract arrays
time = data['time']         # shape: (n_examples, N_PTS)
flux = data['flux']         # shape: (n_examples, N_PTS)
y_coeffs = data['y_coeffs'] # shape: (n_examples, coeff_count)
params = data['parameters'] # shape: (n_examples,)

# Plot the first example's light curve
plt.figure()
plt.plot(time[20], flux[20])
plt.xlabel('Time')
plt.ylabel('Flux')
plt.title('Synthetic Light Curve - Example 20')
plt.show()

# Visualize the corresponding image (if available)
if 'image' in data:
    img = data['image'][20]
    plt.figure()
    plt.imshow(img, origin='lower')
    plt.title('Rendered Map - Example 20')
    plt.axis('off')
    plt.show()
else:
    print('No image data found.')
```

```python
# Cell: Visualize a selection of examples by index
import matplotlib.pyplot as plt

def show_examples(indices):
    """
    Plot light curves and rendered images for the given list of example indices.
    Assumes `time`, `flux`, and (optionally) `data['image']` are already loaded.
    """
    for idx in indices:
        # Plot light curve
        plt.figure()
        plt.plot(time[idx], flux[idx])
        plt.xlabel('Time')
        plt.ylabel('Flux')
        plt.title(f'Light Curve – Example {idx}')
        plt.show()

        # Plot image if available
        if 'image' in data:
            img = data['image'][idx]
            plt.figure()
            plt.imshow(img, origin='lower', cmap='plasma')
            plt.title(f'Rendered Map – Example {idx}')
            plt.axis('off')
            plt.show()
        else:
            print(f"No image data for example {idx}.")

# Example usage:
indices_to_view = [0, 3, 7]
show_examples(indices_to_view)
```
