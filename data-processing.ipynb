```python
# load data from hatp11keplersc.csv using pandas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load data without header
data = pd.read_csv('hatp11keplersc.csv', header=None)
data.columns = ['BJD - 2454833', 'Corrected Flux']
data.head()
```

Kepler light curve of HAT-P-11 - columns are time (Barycentric julian date - 2454833) and flux (median normalized to 1).

```python
data.shape, data.columns
```

```python
# plot the data
plt.figure(figsize=(10, 5))
plt.plot(data['BJD - 2454833'], data['Corrected Flux'])
plt.xlabel('BJD - 2454833')
plt.ylabel('Corrected Flux')
plt.title('HAT-P-11 Kepler Short Cadence')
plt.show()
```
